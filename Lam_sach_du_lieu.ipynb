```python
df = pd.read_csv("jobs.csv")
```

```python
df.isnull().sum()
```

```python
#Xử lý missing

df["skills"] = df["skills"].fillna("Unknown")

df["job_fields"] = df["job_fields"].fillna("Unknown")

df["city"] = df["city"].fillna("Unknown")
```

```python
#Xoá trùng

df = df.drop_duplicates()
```

```python
#Chuẩn hoá text

df["job_title"] = df["job_title"].str.strip()

df["city"] = df["city"].str.strip()
```

```python
#Kiểm tra salary

df[["salary_min","salary_max"]].describe()
```

```python
#Tạo lương trung bình

df["avg_salary"] = (
    df["salary_min"] +
    df["salary_max"]
)/2
```

```python
#Lọc IT Jobs

it_keywords = [
    "developer",
    "software",
    "engineer",
    "backend",
    "frontend",
    "fullstack",
    "data",
    "ai",
    "devops",
    "it"
]
```

```python
pattern = "|".join(it_keywords)

df_it = df[
    df["job_title"]
    .str.lower()
    .str.contains(pattern, na=False)
]
```

```python
#Lưu dữ liệu sạch

df_it.to_csv(
    "clean_it_jobs.csv",
    index=False
)
```
